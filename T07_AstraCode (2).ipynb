```python
## FIltering the data set
filtered_df = df[df['NACCFDYS'] == 0].copy()
```

```python
## Getting all the non medical/diagnosis variables for my model
target_col = 'DEMENTED'
feature_cols = [
    'NACCAGE','EDUC','SEX','HISPANIC','RACE','HANDED','PRIMLANG','MARISTAT','RESIDENC','NACCLIVS','INDEPEND','NACCFAM','NACCMOM','NACCDAD',
    'NACCFFTD','HEIGHT','WEIGHT','NACCBMI','VISION','HEARING','INRELTO','INRELY','NACCREAS','NACCREFR','MMSEVIS','MMSEHEAR','MOCAVIS','MOCAHEAR'
]
```

```python
selected_col = feature_cols + [target_col]
```

```python
final_df = filtered_df[selected_col].copy()
```

```python
#Final Dataset after geting required features and filtering
final_df.head(10)
```

```python
##Using shape funcion getting the rows and cols in the data frame
final_df.shape
```

```python
##Getting the required info and exploring the data frame
final_df.info()
```

```python
final_df.columns
```

```python
##CHecking is there any duplicate values
final_df.duplicated().sum()
```

```python
##Drop the duplicate values
final_df = final_df.drop_duplicates()
```

```python
final_df.shape
```

```python
##Checking the unique values in the features
final_df.nunique()
```

```python
##Handling the features with unknown and Not Available
## Special Codes that represent missing values
special_codes = [-4, 9, 99, 88, 88.8, 888.8, 999]
for col in final_df.columns:
    final_df[col] = final_df[col].replace(special_codes, np.nan)
```

```python
##Checking the missing values
final_df.isnull().sum()/len(final_df)*100
```

```python
##CHecking is there any missing values now
final_df.isnull().sum()
```

```python
## Dropping the columns due to high missing values
final_df = final_df.drop(columns=['MMSEVIS', 'MMSEHEAR'])
```

```python
## Here some of the categorical datas were presented as numerical data so we are gonna deal with them first
categorical_cols = ['SEX', 'HISPANIC', 'RACE', 'HANDED', 'PRIMLANG', 'MARISTAT', 'RESIDENC', 'NACCLIVS', 'INDEPEND','NACCFAM', 'NACCMOM',
                    'NACCDAD', 'NACCFFTD', 'VISION', 'HEARING', 'INRELTO', 'INRELY','NACCREAS', 'NACCREFR', 'MOCAVIS', 'MOCAHEAR']

for cols in categorical_cols:
  final_df[cols] = final_df[cols].astype('object')
```

```python
##Checking the caegorical features data type
final_df.info()
```

```python
## Here we are going to treat EDUC as numerical varibale because it has real continuous meaning so i am gonna treat that as numerical feature
numerical_cols = ['NACCAGE','EDUC','HEIGHT','WEIGHT','NACCBMI']

# let's impute them with the MEDIAN
for col in numerical_cols:
    median_val = final_df[col].median()
    final_df[col] = final_df[col].fillna(median_val)
```

```python
##Lets impute categorical missing values by mode
for col in categorical_cols:

        mode_val = final_df[col].mode()[0]

        final_df[col] = final_df[col].fillna(mode_val)
```

```python
final_df.isnull().sum()
```

```python
##Analysing the target variable
final_df['DEMENTED'].value_counts()
```

```python
##Visualizing the target varibale
##Understanding: Imbalanced data
sns.countplot(x='DEMENTED', data=final_df)
```

```python
##Visualizing the categorical data to get the understing about the categorical data
for col in categorical_cols:
    if col in final_df.columns:
        plt.figure(figsize=(8,4))
        sns.countplot(x=col, data=final_df)
        plt.title(f"{col}")
        plt.show()
```

```python
##Visualizing the categorical features vs traget feature
for col in categorical_cols:
    if col in final_df.columns:
        plt.figure(figsize=(8,4))
        sns.countplot(x=col, hue='DEMENTED', data=final_df)
        plt.title(f"{col} distribution by DEMENTED")
        plt.show()
```

```python
##Seeing the unique values of education col
final_df['EDUC'].unique()
```

```python
##Analysing the numerical feature
for col in numerical_cols:
  data= final_df.copy()
  data[col].hist(bins=25)
  plt.xlabel(col)
  plt.ylabel("Count")
  plt.title(col)
  plt.show()
```

```python
# Plot the distribution of each numerical feature, separated by the target variable 'DEMENTED'.
# This allows us to visually compare how the feature values differ between demented (1) and non-demented (0) subjects.
for col in numerical_cols:
    plt.figure(figsize=(6,4))
    sns.histplot(data=final_df, x=col, hue='DEMENTED', kde=True, bins=30)
    plt.title(f"{col} distribution by DEMENTED")
    plt.show()
```

```python
##Visualizing numerical features using box plot to get the understanding about the outliers
for feature in numerical_cols:
    plt.figure(figsize=(6, 4))

    # Just plot the original data
    sns.boxplot(x=final_df[feature])

    plt.title(f'Boxplot of {feature}')
    plt.xlabel(feature)
    plt.show()
```

```python
#Checking the unique values if the categorical data for the one-hot encoding
for features in categorical_cols:
  print('The feature is {} and number of the categories are {}'.format(features,len(final_df[features].unique())))
```

```python
##One-Hot Encoding
final_df = pd.get_dummies(final_df, columns=categorical_cols,drop_first=True, dtype=int)
```

```python
final_df.head()
```

```python
## Initialysing the dependant and independant features
X = final_df.drop(columns=['DEMENTED'])
y = final_df['DEMENTED']
```

```python
#Train and Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=42,stratify=y)
```

```python
##from imblearn.over_sampling import SMOTE
##smote = SMOTE(random_state=42)
##X_train, y_train = smote.fit_resample(X_train, y_train)
##I used SMOOTE bcz of the imbalance data but it did not help me with the model performace so i commented that
```

```python
#Model Training using random forest
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier()
```

```python
#Train the Data
rf.fit(X_train,y_train)
```

```python
y_pred = rf.predict(X_test)
```

```python
#Model Evaluation
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
acc_score = accuracy_score(y_test,y_pred)
print(acc_score)
```

```python
# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
```

```python
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
# Detailed Report (Precision, Recall, F1)
print("Classification Report:\n", classification_report(y_test, y_pred))
```

```python
#TO evaluate : ROC/AUC curve and score
from sklearn.metrics import roc_curve, auc, roc_auc_score
y_prob = rf.predict_proba(X_test)[:, 1]
fpr, tpr, thresholds = roc_curve(y_test, y_prob)
roc_auc = auc(fpr, tpr)
print("ROC-AUC Score:", roc_auc_score(y_test, y_prob))
```

```python
## Visualizing the ROc curve
plt.figure(figsize=(7,6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label='ROC curve (AUC = %0.2f)' % roc_auc)
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend(loc="lower right")
plt.grid(True)
plt.show()
```

```python
#Importing RandomsizedsearchCV to get the best model Hyper parameters
## We used RandomizedSearchCV insted of GridSearchCV due to the processing time
from sklearn.model_selection import RandomizedSearchCV
```

```python
param_grid = {
    'n_estimators': [100, 200, 300, 400],
    'max_depth': [3, 4, 5, 6, 7],
    'learning_rate': [0.01, 0.05, 0.1, 0.2],
    'subsample': [0.6, 0.8, 1],
    'colsample_bytree': [0.6, 0.8, 1],
    'gamma': [0, 1, 5],
    'min_child_weight': [1, 3, 5]
}
```

```python
xgb_model = xgb.XGBClassifier(eval_metric='logloss')
```

```python
random_search = RandomizedSearchCV(
    estimator=xgb_model,
    param_distributions=param_grid,
    n_iter=30,              # number of different combinations to try
    scoring='roc_auc',      # choose best model by AUC
    cv=5,                   # 5-fold cross-validation
    verbose=2,
    random_state=42,
    n_jobs=-1
)

random_search.fit(X_train, y_train)
```

```python
##Best hyper parameters
print("Best Parameters:", random_search.best_params_)
```

```python
##Best model for the prediction
best_model = random_search.best_estimator_
```

```python
y_pred = best_model.predict(X_test)
y_pred_proba = best_model.predict_proba(X_test)[:, 1]

print("ROC AUC:", roc_auc_score(y_test, y_pred_proba))
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```
